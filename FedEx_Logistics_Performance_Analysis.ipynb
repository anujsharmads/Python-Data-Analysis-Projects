`````python
# droping missing values from important columns and creating its copy
df_clean = df_copy.dropna(subset=[
    'Weight (Kilograms)',
    'Freight Cost (USD)',
    'Line Item Value',
    'Unit Price',
    'Pack Price'
]).copy() #first we have removed those rows which contained the null values for outlier detection

# columns where you want to detect outliers
cols = ['Freight Cost (USD)', 'Unit Price', 'Line Item Value']

#creating a mask (this avoids aggressive row deletion)
mask = pd.Series(True, index=df_clean.index) # True for all rows or Initially, every row is allowed.

for col in cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask &= df_clean[col].between(lower, upper)

#applying filtering once
df_clean = df_clean[mask]
`````

### Outlier Detection and Treatment Workflow

The following workflow was used for outlier detection and treatment in the dataset:

````
Original DataFrame
        │
        ▼
Remove rows with missing values using dropna()
        │
        ▼
Create a cleaned DataFrame (df_clean)
        │
        ▼
Select important numerical columns for outlier detection
        │
        ▼
Calculate Q1, Q3, and IQR for each selected column
        │
        ▼
Determine lower and upper bounds using the IQR method
        │
        ▼
Create and update a boolean mask (True = keep, False = outlier)
        │
        ▼
Apply the mask to filter the dataset
        │
        ▼
Generate the final cleaned dataset without the selected outliers
````

### Explanation

1. Missing values from important numerical columns were removed using `dropna()` to ensure accurate outlier detection.

2. A new DataFrame (`df_clean`) was created to preserve the original dataset.

3. Numerical columns such as **Freight Cost (USD)**, **Unit Price**, and **Line Item Value** were selected for outlier analysis.

4. For each column, the first quartile (Q1), third quartile (Q3), and Interquartile Range (IQR) were calculated.

5. The lower and upper limits were computed using the formula:

* Lower Bound = Q1 − 1.5 × IQR
* Upper Bound = Q3 + 1.5 × IQR

6. A boolean mask was created where **True** indicates values within the acceptable range and **False** indicates potential outliers.

7. The mask was updated for each selected column, ensuring that only rows without outliers across all chosen columns were retained.

8. Finally, the mask was applied to obtain a cleaned dataset suitable for further visualization and statistical analysis.
